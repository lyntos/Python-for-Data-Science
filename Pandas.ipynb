Statistics is the branch of mathematics that deals with collecting, organizing, analyzing, interpreting, and presenting data.

In simple terms:
 It helps you understand data and make decisions based on it.

    It does NOT make predictions

Examples:

    Mean (average)

    Median

    Mode

    Standard deviation

Finding the average monthly charges of customers

Used to make predictions or conclusions about a population based on a sample

Examples:

    Confidence intervals

    Regression analysis

 Example:

Predicting whether a customer will churn or not based on sample data

 Easy way to remember:

    Descriptive → “What does the data show?”

Step 5: Descriptive Statistics

    Mean, median, mode:


print(df['NumericColumn'].mean())

print(df['NumericColumn'].mode()[0])

```python
# Here’s an example for MonthlyCharges:

# Mean
print(df['MonthlyCharges'].mean())

# Median
print(df['MonthlyCharges'].median())

# Mode (most frequent value)
print(df['MonthlyCharges'].mode()[0])
```

Key points:

    .mean() → average value

    .median() → middle value

    .mode() → returns a Series; [0] gives the most frequent value

-->You can replace 'MonthlyCharges' with any numeric column like 'tenure' or 'TotalCharges'.

Tip: Make sure your numeric columns are actually float or int (df.dtypes)—otherwise .mean() and .median() won’t work.

# Is it necessary to give [0] in mode()[0]?

Yes — in most cases, you need [0] when using .mode(). Here’s why:

df['NumericColumn'].mode() returns a Series, not a single value.

This Series can have more than one value if there’s a tie for the most frequent value.

[0] picks the first mode in that Series.



```python
# Example:

import pandas as pd

data = [1, 2, 2, 3, 3]
s = pd.Series(data)

print(s.mode())    # Output: 0    2
                   #         1    3
                   # dtype: int64

print(s.mode()[0]) # Output: 2  ← first mode

#  If you don’t use [0] and try to print directly, you’ll get a Series object, not a single number.
```

# What is a Formatted String?

A formatted string (often called an f-string in Python) is a way to insert variables or expressions directly inside a string.

 Example of formatted string:

name = "Alice"

age = 25

print(f"My name is {name} and I am {age} years old")

 Output:

My name is Alice and I am 25 years old




```python
# 1. Normal Print
name = "Alice"
age = 25

print("My name is", name, "and I am", age, "years old")

# Uses commas to join values

# Less control over formatting

# Can look messy for complex output

```

```python
# 2. Formatted Print (f-string)
print(f"My name is {name} and I am {age} years old")

# Cleaner and easier to read

# Allows inserting variables directly

# Supports expressions like {age + 5}
```

```python
# Extra Example (with expression):
print(f"In 5 years, I will be {age + 5}")

# 💡 Simple idea:

# Normal print = joining pieces

# Formatted string = filling in blanks inside a sentence
```

```python
# 1. Using .format() method
name = "Alice"
age = 25

print("My name is {} and I am {} years old".format(name, age))

# 👉 Output:

# My name is Alice and I am 25 years old
```

```python
# 2. Using % formatting (old style)
name = "Alice"
age = 25

print("My name is %s and I am %d years old" % (name, age))
```

# Quick Comparison

    f-string → f"My name is {name}" (modern & best)

    .format() → "My name is {}".format(name)

    % formatting → "My name is %s" % name (old style)

 Best practice today: Use f-strings (simple and readable)

# 1.loc[] — label-based selection

In pandas, loc is used to select data from a DataFrame using labels (row names and column names).

 Simple Definition:

 loc is used to access rows and columns by their names (labels).


Syntax:

df.loc[row_labels, column_labels]

```python
# Examples:
# 1. Select a single row by index label
print(df.loc[0])  # first row (if index starts at 0)

# 2. Select a single column
print(df.loc[:, 'gender'])  # all rows, column 'gender'

# 3. Select multiple columns
print(df.loc[:, ['gender', 'Churn']])

# 4. Select rows by condition
print(df.loc[df['Churn'] == 'Yes', ['customerID', 'MonthlyCharges']])

#  Key point: .loc includes the last label in slicing, unlike Python normal slicing.
```

```python
print(df.loc[0:3, 'gender':'tenure'])  # includes row 3 and column 'tenure'
```

#  .iloc[] — position-based selection

    In pandas, iloc is used to select data using integer positions (index numbers).

Simple Definition:

    iloc is used to access rows and columns by their position (numbers).

 Basic Syntax:

    df.iloc[row_index, column_index]



```python
# Examples:
# 1. Select first row
print(df.iloc[0])

# 2. Select first column
print(df.iloc[:, 0])

# 3. Select multiple rows and columns
print(df.iloc[0:3, 0:2])  # rows 0,1,2 and cols 0,1

# 4. Select last row
print(df.iloc[-1])
```
